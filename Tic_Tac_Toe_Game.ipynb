```python
# Game Project OOP Essentials 1
# TIC-TAC-TOE PROJECT

# Import choice() to allow the computer to make a random move
from random import choice

# Game board
# Numbers represent available positions
# X is the computer's starting move in the center
board = [1, 2, 3, 4, 'X', 6, 7, 8, 9]

# All possible winning combinations
winning_lines = [
    (0, 1, 2), (3, 4, 5), (6, 7, 8),  # Rows
    (0, 3, 6), (1, 4, 7), (2, 5, 8),  # Columns
    (0, 4, 8), (2, 4, 6)              # Diagonals
]

# Display the current game board
def display_board():
    print()
    print("+-------+-------+-------+")
    print(f"|   {board[0]}   |   {board[1]}   |   {board[2]}   |")
    print("+-------+-------+-------+")
    print(f"|   {board[3]}   |   {board[4]}   |   {board[5]}   |")
    print("+-------+-------+-------+")
    print(f"|   {board[6]}   |   {board[7]}   |   {board[8]}   |")
    print("+-------+-------+-------+")
    print()

# Return a list of unoccupied cells
# Any integer on the board represents a free cell
def free_cells():
    return [cell for cell in board if isinstance(cell, int)]

# Handle the player's move
def player_move():
    while True:
        try:
            # Ask the player to choose a position
            move = int(input("Choose a free square (1-9): "))

            # Validate that the square is available
            if move not in free_cells():
                print("That square is unavailable.")
                continue

            # Place player's symbol (O)
            board[move - 1] = 'O'
            break

        except ValueError:
            # Handle non-numeric input
            print("Please enter a valid number.")

# Handle the computer's move
def computer_move():
    # Randomly select one of the available cells
    move = choice(free_cells())

    # Place computer's symbol (X)
    board[move - 1] = 'X'

    print(f"Computer selected: {move}")

# Check if the given symbol has won the game
def winner(symbol):
    # Test every winning combination
    for a, b, c in winning_lines:
        if board[a] == board[b] == board[c] == symbol:
            return True

    return False

# Determine whether the game has ended
def game_over():
    # Check if computer wins
    if winner('X'):
        display_board()
        print("Computer wins!")
        return True

    # Check if player wins
    if winner('O'):
        display_board()
        print("You win!")
        return True

    # Check for draw (no free cells remaining)
    if len(free_cells()) == 0:
        display_board()
        print("It's a draw!")
        return True

    # Game continues
    return False

# Welcome message
print("Welcome to Tic-Tac-Toe!")

# Main game loop
while True:

    # Show the current board
    display_board()

    # Player takes a turn
    player_move()

    # Check if game ended after player's move
    if game_over():
        break

    # Computer takes a turn
    computer_move()

    # Check if game ended after computer's move
    if game_over():
        break
```
